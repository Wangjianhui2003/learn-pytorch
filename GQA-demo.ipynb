```python
import math

# demonstrate QK^T in GQA
torch.manual_seed(123)
Q = torch.rand(2, 4, 5, 10)
K = torch.rand(2, 4, 7, 10)
scores = Q @ K.transpose(-2, -1) / math.sqrt(10)
print(scores.shape, scores)

q_len, k_len = scores.shape[-2], scores.shape[-1]

allowed = torch.ones(q_len, k_len, dtype=torch.bool, device="cpu")
print(allowed)
allowed = torch.tril(allowed, diagonal=k_len - q_len)
print(allowed)
scores = scores.masked_fill(~allowed, torch.finfo(scores.dtype).min)
print(scores)
```
