## Dim collapse

dim specifies which axis is collapsed, not which values you add along visually

"Collapse" = remove that axis and sum over every index along it. The other axes keep their sizes and order.

**1. Start with `[2,3]` to see the two directions**

```
A =  1  2  3        dim=0: sum down each column
     4  5  6
```

```
dim=0                          dim=1
 1  2  3                       1  2  3  ->  6
 |  |  |          collapse     ---------
 v  v  v          rows        4  5  6  -> 15
 4  5  6                        (sum each row)
 --------
 5  7  9   shape [3]           shape [2] (or [2,1] with keepdim)
```

`dim=0` = arrows pointing **down the axis-0 direction**; that axis vanishes. `dim=1` = arrows pointing **across the last axis**.

**2. `[8,4,3,10]` as nested boxes**

Think of it as 8 boxes, each box holds 4 sheets, each sheet is `3x10`:

```
dim=0  (stack the 8 boxes, add corresponding elements)
  8 boxes:  [B0] [B1] ... [B7]
               \   |      /
                elementwise +
                    |
              one box [4,3,10]
```



For `A` of shape `[8,4,3,10]`, with indices `A[i,j,k,l]`:

| call | output shape | element computed |
|---|---|---|
| `sum(dim=0)` | `[4,3,10]` | `out[j,k,l] = Σ_i A[i,j,k,l]` |
| `sum(dim=1)` | `[8,3,10]` | `out[i,k,l] = Σ_j A[i,j,k,l]` |
| `sum(dim=2)` | `[8,4,10]` | `out[i,j,l] = Σ_k A[i,j,k,l]` |
| `sum(dim=3)` (`-1`) | `[8,4,3]` | `out[i,j,k] = Σ_l A[i,j,k,l]` |

`keepdim=True` does the same reduction but leaves a `1` in the collapsed slot: `dim=1` → `[8,1,3,10]`, so broadcasting still works.

Multiple axes: `sum(dim=(0,2))` removes both → `[4,10]`, computing `out[j,l] = Σ_i Σ_k A[i,j,k,l]`.

The mental rule: **the reduced dim disappears; all other dims are just "batch" axes passed through unchanged**, and each output element is the sum over the full slice along the collapsed dim.


**softmax does not collapse — it normalizes along that axis and returns the same shape**
