```
[[fill]align][sign][#][0][width][,][.precision][type]
```
[how to pad] [how to sign] [how wide] [how to group] . [how many decimals] [what kind]

````python
f"{3.14159:.3f}"
````

````python
f"{1234.5678:,}"
````

````python
f"{2.5:03f}"  # 3 mean minimum number. fixed-point, default precision = 6
````

````python
f"{2.5:07.03f}"  # width is 7, include . sign
````

````python
f"{0.257:.1%}"
````

````python
x = -3.14159
f"{x:>7.2f}", f"{x:#>7.2f}"
````

````python
# Write the format spec that prints 255 as lowercase hex with a 0x prefix,
# zero-padded to a total width of 6 characters. (Expected output: 0x00ff.)
f"{255:#06x}", f"{255:x}", f"{255:#x}", f"{255:#X}", f"{255:#b}"
````

Write the format spec that prints 42 as an integer, zero-padded to width 6. (Expected output: 000042.)

````python
f"{42:06d}", f"{42:06,d}"
````

````python
# f"{3.14159:.2f}" vs f"{3.14159:2f}" — what does each print, and what is the 2 doing in each case?
f"{3.14159:.2f}", f"{3.14159:2f}"  # width 2, which is exceeded and therefore ignored; default 6 decimals
````

````python
# Decompose f"{x:+.2e}" piece by piece (what is each character's job?), and give the output for x = 1234.5.
x = 1234.5
f"{x:+.2e}"  # The exponent is always signed and zero-padded to at least 2 digits
````

````python
# Given w = 8, p = 3, x = 1.23456, what does f"{x:{w}.{p}f}" print? Count the spaces.
w = 8
p = 3
x = 1.23456
f"{x:{w}.{p}f}"
````
