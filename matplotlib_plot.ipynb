```python
import matplotlib.pyplot as plt
import torch
```

```python
# draw a func
x = torch.linspace(-3, 3, 100)
y = x**2

# plt.subplot
ax1 = plt.subplot(1, 2, 1)  # state-based api
plt.plot([1, 2, 3, 4, 5], [1, 2, 5, 4, 5])  # state-based api

fig = ax1.get_figure()
fig.set_size_inches(8, 3)
# fig = plt.gcf()
# fig.set_size_inches(8,3)

# All three calls use gcf() (get current figure)
# and a figure already exists after the first call
# so no new figure is created
ax2 = plt.subplot(1, 3, 1)
ax3 = plt.subplot(1, 4, 1)

# ax1.remove()
# ax2.remove()
ax3.remove()

fig2, ax4 = plt.subplots(2, 2, figsize=(10, 4))

ax4[0][0].plot(x, y)
plt.show()
```

```python
# plt.subplots
fig, axs = plt.subplots(2, 2)  # 1 fig, 2x2 array axes
axs[0][0].plot([1, 2, 3], [1, 2, 3])
axs[0][1].set_title("the 2nd axes")
fig.set_size_inches(12, 8)
fig.tight_layout()
```

```python
fig1 = plt.figure(figsize=(8, 4))  # create and change current figure (optional to return value)
plt.plot([2, 4, 5, 1], [4, 5, 21, 3])
fig2 = plt.figure(figsize=(8, 5))  # create and change current figure
plt.plot([2, 4, 5, 1], [4, 5, 21, 4])
```

```python
fig = plt.figure(figsize=(8, 4))  # fig, no axes yet
ax = fig.add_subplot(111)  # now you have something to plot on
ax.plot(x, y)
```
