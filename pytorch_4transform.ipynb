```python
import torch
import torch.nn.functional as F
from torchvision import datasets
from torchvision.transforms import v2

# Data does not always come in its final processed form that is required for training machine learning algorithms. We use transforms to perform some manipulation of the data and make it suitable for training.

# All TorchVision datasets have two parameters -transform to modify the features and target_transform to modify the labels - that accept callables containing the transformation logic. 

ds = datasets.FashionMNIST(
    root="data",
    train=True,
    download=True,
    transform=v2.Compose([v2.ToImage(), v2.ToDtype(torch.float32, scale=True)]),
    target_transform=v2.Lambda(
        lambda y: F.one_hot(torch.tensor(y), num_classes=10).float()
    ),
)

# They do the following:
# 1. v2.ToImage() converts the input into a torchvision image tensor. It does not change the value range.
# 2. v2.ToDtype(torch.float32, scale=True) converts the tensor to 32-bit floating point and scales integer pixel values, typically from 0..255 to 0.0..1.0.
```
