## Linear Layer

`x @ W.T + b` — with `x` shape `(..., in_features)` and `weight` shape `(out_features, in_features)`. Equivalently, treating `x` as a column vector: `W @ x + b`. The stored weight is transposed relative to the math convention, so it's **not** `x @ W`.

```python
import torch
import torch.nn as nn
```

```python
linear = nn.Linear(4, 5)
linear.weight
```

```python
linear.bias
```

```python
linear.weight.T
```

```python
t1 = torch.ones(3,4)
t1
```

```python
linear(t1)
```

```python
t1 @ linear.weight.T + linear.bias
```

```python
t2 = torch.ones(4)
t2
```

```python
t2 @ linear.weight.T + linear.bias 
```

```python
linear.weight @ t2 + linear.bias # treat t2 as a column vector
```
