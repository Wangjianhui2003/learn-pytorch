Any tensor that is **module state but not a trainable parameter** should be a buffer, so `.to()/.cuda()/.float()/.half()` and `state_dict()` handle it automatically:

- BN stats: `running_mean`, `running_var`, `num_batches_tracked`
- Fixed masks (causal/attention masks), positional encodings
- Constant scale factors, class weights, lookup tables, fixed embeddings
- Accumulators like `total_steps` you want saved/moved

If you just assign a raw tensor as an attribute (`self.mask = torch.tensor(...)`), PyTorch won't move or save it — you must call `.to(device)` yourself. Use `self.register_buffer("mask", t)` (or `persistent=False` to keep it out of `state_dict`).
